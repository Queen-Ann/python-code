### PATIENTS FILE DETAILS

```python
def patients_demo_info(first_name, last_name, **kwargs):
    """Print out patient details including additional informations."""
    print(f'Patient Name: {first_name} {last_name}')
    for key, value in kwargs.items():
        print(f'{key.capitalize()}: {value}')
        
def patient_department_info(**kwargs):
    """Display the department name and other details."""
    department = kwargs.get('department')
    print(f"Patient Department is {department}")
    for key, value in kwargs.items():
        if key != 'department':  # Skip department if it's already printed
            print(f"{key}: {value}")
            
def patient_health_info(condition, treatment):
    """Print the patient’s health condition and treatment."""
    if condition:
        print(f'Patient condition is {condition}')
    else:
        print('No condition provided.')
    if treatment:
        print(f'Treatment plan is {treatment}')
    else:
        print('No treatment provided.')
```

```python
# The function Call.
#Patient demographic informations details
patients_demo_info("Ann", "Smith", age=28, gender="female", nationality="Nigeria", date="2024-08-13",
              patient_id=456, Address="block 5 Cent_city_Enugu", phone_no="08098765431")

#printing out the department name, the doctor incharge, the admission status, the ward and bed number
patient_department_info(department='Neurology', doctor='Dr_Brian', admitted='yes', Ward=101, Bed=10)

#printing the patient's health condition and treatmemt plan
patient_health_info('Dementia', 'lifestyle changes and cognitol')
```
